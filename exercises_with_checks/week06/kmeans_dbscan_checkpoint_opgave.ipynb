## Oefening 3 — KMeans vs DBSCAN op Iris

De Iris-dataset bevat 3 soorten die in de feature-ruimte goed gescheiden, **bolvormige** clusters van **gelijkaardige dichtheid** vormen.

Pas zowel KMeans (verschillende `k`) als DBSCAN (verschillende `eps`/`min_samples`) toe op de **gestandaardiseerde** data en vergelijk met de silhouette score. Bewaar je gekozen `k` in `best_k_iris` en de hoogste silhouette in `silhouette_iris`. Beantwoord daarna vraag 1 in `antwoord_iris`.

```python
import numpy as np
from sklearn.datasets import load_iris
from sklearn.cluster import KMeans, DBSCAN
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import silhouette_score

iris = load_iris()
data_scaled = StandardScaler().fit_transform(iris.data)

# TODO: pas KMeans toe voor k in range(2, 6) en DBSCAN voor een aantal
# (eps, min_samples) combinaties, en vergelijk de silhouette scores.
# Bewaar de gekozen k in best_k_iris en de hoogste silhouette in silhouette_iris.
best_k_iris = 2
silhouette_iris = 0.0

print("--- KMEANS ---")
for k in range(2, 6):
    # Model maken, direct fitten en labels uitdelen
    model_km = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels_km = model_km.fit_predict(data_scaled)
    
    # Rapportcijfer berekenen
    score_km = silhouette_score(data_scaled, labels_km)
    print(f"k={k} heeft score: {score_km:.4f}")
    
    # Bewaar de allerbeste score voor de checkpoint onderaan je notebook
    if score_km > silhouette_iris:
        silhouette_iris = score_km
        best_k_iris = k

# === DEEL 2: DBSCAN TESTEN ===
print("\n--- DBSCAN ---")
eps_opties = [0.4, 0.5, 0.6]
min_samples_opties = [3, 4, 5]

for e in eps_opties:
    for m in min_samples_opties:
        model_db = DBSCAN(eps=e, min_samples=m)
        labels_db = model_db.fit_predict(data_scaled)
        
        # Trucje: Silhouette score crasht als er maar 1 groep wordt gevonden
        # We tellen hier de unieke groepen (zonder de ruis '-1' mee te rekenen)
        aantal_groepen = len(set(labels_db)) - (1 if -1 in labels_db else 0)
        
        if aantal_groepen > 1:
            score_db = silhouette_score(data_scaled, labels_db)
            print(f"eps={e}, min_samples={m} | Score: {score_db:.4f} ({aantal_groepen} clusters)")
```

## Oefening 4 — KMeans vs DBSCAN op Mall

De Mall-dataset (`data/Mall_Customers.csv`) bevat klanten met `Age`, `Annual Income (k$)` en `Spending Score (1-100)`. De clusters hebben **sterk verschillende dichtheden**, waardoor DBSCAN het vaak moeilijk heeft.

Pas KMeans (verschillende `k`) en DBSCAN (verschillende `eps`/`min_samples`) toe op de **gestandaardiseerde** data en vergelijk met de silhouette score. Bewaar je gekozen `k` in `best_k_mall` en de hoogste silhouette in `silhouette_mall`. Beantwoord daarna vraag 2 in `antwoord_dbscan`.

```python
import numpy as np
import pandas as pd
from sklearn.cluster import KMeans, DBSCAN
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import silhouette_score

mall = pd.read_csv('data/Mall_Customers.csv')
features = mall[['Age', 'Annual Income (k$)', 'Spending Score (1-100)']]
data_scaled = StandardScaler().fit_transform(features)

# TODO: vergelijk KMeans (verschillende k's) met DBSCAN (verschillende
# eps/min_samples) via de silhouette score. Bewaar de gekozen k in best_k_mall
# en de hoogste silhouette in silhouette_mall.
best_k_mall = 5
silhouette_mall = 0.0

print("--- KMEANS ---")
for k in range(2, 8):
    # Model maken, fitten en labels uitdelen
    model_km = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels_km = model_km.fit_predict(data_scaled)
    
    # Bereken het rapportcijfer
    score_km = silhouette_score(data_scaled, labels_km)
    print(f"k={k} heeft score: {score_km:.4f}")
    
    # Bewaar de allerbeste score voor de automatische nakijk-check
    if score_km > silhouette_mall:
        silhouette_mall = score_km
        best_k_mall = k

print(f"\nWinnaar KMeans: k={best_k_mall} (Score: {silhouette_mall:.4f})\n")


# === DBSCAN OP DE MALL DATA ===
print("--- DBSCAN ---")
# We testen verschillende cirkelgroottes (eps) en minimum groepsgroottes
eps_opties = [0.3, 0.5, 0.7, 0.9]
min_samples_opties = [3, 5, 7]

for e in eps_opties:
    for m in min_samples_opties:
        model_db = DBSCAN(eps=e, min_samples=m)
        labels_db = model_db.fit_predict(data_scaled)
        
        # Veiligheidscheck: we tellen het aantal groepen (ruis '-1' telt niet mee als groep)
        aantal_groepen = len(set(labels_db)) - (1 if -1 in labels_db else 0)
        
        # Silhouette score kan alleen berekend worden als er minstens 2 groepen zijn
        if aantal_groepen > 1:
            score_db = silhouette_score(data_scaled, labels_db)
            print(f"eps={e}, min_samples={m} | Score: {score_db:.4f} ({aantal_groepen} clusters)")
```
